```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
data=pd.read_csv('Churn_Modelling.csv')
```

```python
data.head()
```

```python
print(data.columns)
```

```python

data
```

```python
#encoding categorical data
labelencoder_gender=LabelEncoder()
data['Gender']=labelencoder_gender.fit_transform(data['Gender'])
data
```

```python
#one hot encoding for geography column
from sklearn.preprocessing import OneHotEncoder
onehotencoder=OneHotEncoder()
geo=onehotencoder.fit_transform(data[['Geography']])
geo
```

```python
onehotencoder.get_feature_names_out(['Geography'])
```

```python
geo_df=pd.DataFrame(geo.toarray(),columns=onehotencoder.get_feature_names_out(['Geography']))
geo_df
```

```python
#combine
data=pd.concat([data,geo_df],axis=1)
```

```python
data=data.drop(['Geography'],axis=1)
```

```python
data.head()
```

```python
###save the encoder
with open('labelencoder_gender.pkl','wb') as file:
    pickle.dump(labelencoder_gender,file)
with open('onehotencoder.pkl','wb') as file:
    pickle.dump(onehotencoder,file)
```

```python
# Divinding dataset into dependent and independent variables
X=data.drop(['Exited'],axis=1)
y=data['Exited']

##Split in training and testing data
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)

##Scale these features
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test) 

```

```python
X_train
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime

```

```python
#BUilding the ANN model
model=Sequential([
    Dense(64, activation='relu',input_shape=(X_train.shape[1],)),#first hidden layer
    Dense(32, activation='relu'),#second hidden layer
    Dense(1, activation='sigmoid')#output layer
])
```

```python
model.summary()
```

```python
##compile the model
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
 ##setup the tensorboard

from tensorflow.keras.callbacks import TensorBoard,EarlyStopping


log_dir="logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
#early stopping to avoid overfitting
early_stopping=EarlyStopping(monitor='val_loss', patience=10,   restore_best_weights=True)
```

```python
##Train the model
history=model.fit(X_train,y_train,validation_data=(X_train, y_train),epochs=100,batch_size=32,callbacks=[tensorflow_callback,early_stopping])
```

```python
model.save('model.h5')
```
