```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import pandas as pd
import pickle
```

```python
#load the model, scaler and label encoder
model = load_model('model.h5')
with open('labelencoder_gender.pkl', 'rb') as f:
    label_encoder = pickle.load(f)
      
with open('onehotencoder.pkl', 'rb') as f:
    label_encoder_geo = pickle.load(f)

with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
```

```python
input_data ={
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Female', 
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000, 
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encoded = label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_df = pd.DataFrame(geo_encoded, columns=label_encoder_geo.get_feature_names_out(['Geography']))
geo_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
#encode categorical variables
input_df['Gender'] = label_encoder.transform(input_df['Gender'])
input_df
```

```python
#concatenate thhe one hot encoded columns with the input dataframe
input_df = pd.concat([input_df.drop('Geography', axis=1), geo_df], axis=1)
```

```python
#scaling the inpput data
scaled_input = scaler.transform(input_df)
```

```python
scaled_input    
```

```python
#predict churn
prediction = model.predict(scaled_input)
```

```python
prediction
```

```python
prediction_proba = prediction[0][0]
prediction_proba
```

```python
if prediction_proba > 0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is unlikely to churn.")
```
